```python
# First 4 verification columns
checks = list(df.columns[2:6])

# Check whether phone number is available
df["Phone Available"] = df[df.columns[6]].notna().map({
    True: "Y",
    False: "N"
})

# Add phone availability to our checks
checks.append("Phone Available")

# Calculate verification score
df["Verification Score"] = df[checks].eq("Y").sum(axis=1)

print(df[["Member Name", "Verification Score"]])
```

```python
import matplotlib.pyplot as plt
```

```python
df["Sub division"].value_counts().plot(kind="bar")

plt.title("Beneficiaries by Sub Division")
plt.xlabel("Sub Division")
plt.ylabel("Number of Beneficiaries")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["1 Oct Avalilability"].value_counts().plot(kind="bar")

plt.title("1 October Availability")
plt.xlabel("Availability")
plt.ylabel("Number of Beneficiaries")
plt.tight_layout()
plt.show()
```

```python
df["Remarks"].value_counts().plot(kind="bar")

plt.title("Beneficiary Status / Remarks")
plt.xlabel("Remarks")
plt.ylabel("Number of Beneficiaries")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
